## 🛠️ Configuración del Entorno QIIME 2

Antes de comenzar con los análisis, necesitamos configurar **QIIME 2 2025.10** en Google Colab. Este proceso incluye:

1. **Instalación de Miniconda**

3. **Instalación de paquetes adicionales** como `empress` para visualizaciones
4. **Verificación** de que todo funciona correctamente

### ⚠️ Importante
- Este proceso puede tardar **10-15 minutos** la primera vez
- Es normal ver mucha salida de texto durante la instalación
- **No interrumpas** el proceso mientras se está ejecutando

```python
# 🚀 Paso 1: Ejecutar el script de configuración de QIIME 2
# Este script instalará y configurará todo lo necesario para usar QIIME 2 2025.10

import os
import subprocess
from pathlib import Path

# Navegar al directorio del repositorio clonado
repo_path = "/content/Microbioma_Amplicones_16S_ITS_QIIME2_2025.10"

if os.path.exists(repo_path):
    os.chdir(repo_path)
    print(f"📁 Cambiado al directorio: {os.getcwd()}")
else:
    print("❌ Error: No se encontró el directorio del repositorio")
    print("   Asegúrate de ejecutar primero la celda de clonado")

# Verificar que el script existe
script_path = "scripts/setup_qiime2.sh"
if os.path.exists(script_path):
    print(f"✅ Script encontrado: {script_path}")
    print("🔄 Iniciando instalación de QIIME 2...")
    print("   Este proceso puede tardar 10-15 minutos...")
else:
    print(f"❌ Error: No se encontró el script {script_path}")
    print("   Verifica que el repositorio se clonó correctamente")
```

```python
# 🔧 Paso 2: Ejecutar el script de instalación
# ⚠️ ATENCIÓN: Esta celda puede tardar 10-15 minutos en ejecutarse
# No interrumpas el proceso mientras se está ejecutando

!bash scripts/setup_qiime2.sh
```

```python
# ✅ Paso 3: Activar QIIME 2 y verificar instalación
# Activamos el entorno de QIIME 2 y verificamos que todo funciona correctamente

import subprocess
import sys

print("🔄 Activando entorno QIIME 2...")

# Activar el entorno QIIME 2
activation_commands = [
    "source /opt/miniconda/etc/profile.d/conda.sh",
    "conda activate qiime2-2025.10",
    "qiime info"
]

# Ejecutar los comandos
command = " && ".join(activation_commands)
result = subprocess.run(f"bash -c '{command}'", shell=True, capture_output=True, text=True)

if result.returncode == 0:
    print("✅ QIIME 2 instalado y funcionando correctamente!")
    print("\n📋 Información del sistema QIIME 2:")
    print(result.stdout)
else:
    print("❌ Error al activar QIIME 2:")
    print(result.stderr)
    print("\n💡 Sugerencias:")
    print("   1. Verifica que el script de instalación se ejecutó completamente")
    print("   2. Revisa los mensajes de error anteriores")
    print("   3. Intenta ejecutar nuevamente el script de instalación")
```

## 📂 Preparación de Datos

Ahora que QIIME 2 está instalado, vamos a preparar la estructura de datos para tus análisis.

### Estructura de carpetas disponibles:

- **`data_example/16S/`** 🦠: Para datos de amplicones de 16S rRNA (bacterias y arqueas)
- **`data_example/ITS/`** 🍄: Para datos de amplicones de ITS (hongos)

Cada carpeta contiene:
- Archivos README con instrucciones específicas
- Datos de ejemplo para practicar
- Espacio para tus propios datos

```python
# 📋 Explorar estructura de datos disponibles
import os
from pathlib import Path

print("🔍 Explorando la estructura de datos disponibles...\n")

# Listar contenido de las carpetas de datos
data_folders = ["data_example/16S", "data_example/ITS"]

for folder in data_folders:
    if os.path.exists(folder):
        print(f"📁 {folder}/")
        files = os.listdir(folder)
        if files:
            for file in sorted(files):
                file_path = os.path.join(folder, file)
                if os.path.isfile(file_path):
                    size = os.path.getsize(file_path)
                    size_mb = size / (1024 * 1024)
                    print(f"   📄 {file} ({size_mb:.1f} MB)")
                elif os.path.isdir(file_path):
                    print(f"   📂 {file}/")
        else:
            print("   📝 Carpeta vacía - lista para tus datos")
        print()
    else:
        print(f"❌ No se encontró la carpeta: {folder}")

print("💡 Consejos:")
print("   • Usa data_example/16S/ para datos de bacterias y arqueas")
print("   • Usa data_example/ITS/ para datos de hongos")
print("   • Lee los archivos README.md en cada carpeta para más información")
```
