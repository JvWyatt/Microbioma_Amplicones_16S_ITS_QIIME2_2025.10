# 🛠️ Instalar QIIME 2

QIIME 2 es la herramienta principal para analizar microbiomas.

**¿Qué hace?** Procesa secuencias de ADN para identificar y estudiar microorganismos (bacterias 🦠, hongos 🍄).

Se instalará **QIIME 2 2025.10 Amplicon Distribution** siguiendo las [instrucciones oficiales](https://library.qiime2.org/quickstart/amplicon).

⏱️ **Tiempo estimado**: 5-10 minutos

### Paso 1: Configurar Miniconda (2-3 minutos)
Instala y configura Conda, el gestor de paquetes necesario para QIIME 2.

# 📦 Instalar y configurar Conda

```python
import subprocess
import sys
import os
from IPython.display import clear_output
import time

os.chdir('/content/Microbioma_Amplicones_16S_ITS_QIIME2_2025.10')

print("📦 Instalando Miniconda...")
print("   Por favor espera, esto tomará 2-3 minutos\n")

# Ejecutar instalación en silencio
process = subprocess.Popen(
    ['bash', 'scripts/setup_conda.sh'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    universal_newlines=True,
    cwd='/content/Microbioma_Amplicones_16S_ITS_QIIME2_2025.10'
)

# Mostrar animación de progreso
spinner = ['⠋', '⠙', '⠹', '⠸', '⠼', '⠴', '⠦', '⠧', '⠇', '⠏']
i = 0
while process.poll() is None:
    print(f"\r   {spinner[i % len(spinner)]} Configurando Conda...", end='', flush=True)
    time.sleep(0.1)
    i += 1

stdout, stderr = process.communicate()

print("\r", end='')
if process.returncode == 0:
    print("✅ Conda instalado y configurado correctamente")
    print("   Versión instalada: Miniconda (última versión)")
else:
    print("❌ Error durante la instalación")
    print("\nDetalles del error:")
    print(stderr)
```

#### ✅ Verificar instalación de Conda
Comprueba que Miniconda se instaló correctamente y muestra la versión.

```python
# 🔍 Verificar instalación de Conda
import subprocess

print("🔍 Verificando instalación...")

# Verificar versión de Conda
result = subprocess.run(
    ['conda', '--version'],
    capture_output=True,
    text=True,
    env={'PATH': '/opt/miniconda/bin:/usr/local/bin:/usr/bin:/bin'}
)

if result.returncode == 0:
    version = result.stdout.strip()
    print(f"✅ {version}")
    print("   Conda está listo para usar\n")
else:
    print("❌ Conda no está disponible")
    print("   Por favor, ejecuta la celda anterior de nuevo\n")
```

### Paso 2: Instalar QIIME 2 (5-10 minutos)
Instala QIIME 2 2025.10 Amplicon Distribution con todos sus componentes.

```python
# 🧬 Instalar QIIME 2 2025.10 Amplicon
import subprocess
import sys
import os
import time

os.chdir('/content/Microbioma_Amplicones_16S_ITS_QIIME2_2025.10')

print("🧬 Instalando QIIME 2 2025.10 Amplicon Distribution...")
print("   Este proceso puede tomar 5-10 minutos\n")

# Ejecutar instalación
process = subprocess.Popen(
    ['bash', 'scripts/install_qiime2.sh'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    universal_newlines=True,
    cwd='/content/Microbioma_Amplicones_16S_ITS_QIIME2_2025.10'
)

# Mostrar progreso
spinner = ['⠋', '⠙', '⠹', '⠸', '⠼', '⠴', '⠦', '⠧', '⠇', '⠏']
i = 0
start_time = time.time()

while process.poll() is None:
    elapsed = int(time.time() - start_time)
    mins = elapsed // 60
    secs = elapsed % 60
    print(f"\r   {spinner[i % len(spinner)]} Instalando QIIME 2... ({mins}m {secs}s)", end='', flush=True)

    time.sleep(0.1)        print(stderr[-500:])  # Últimas 500 caracteres del error

    i += 1    if stderr:

    print("\nDetalles del error:")

stdout, stderr = process.communicate()    print("❌ Error durante la instalación")

else:

print("\r", end='')    print(f"   Tiempo total: {elapsed // 60}m {elapsed % 60}s")

if process.returncode == 0:    print(f"✅ QIIME 2 2025.10 instalado correctamente")
    elapsed = int(time.time() - start_time)
```

```python
# ✅ Verificar instalación de QIIME 2
import subprocess

print("🔍 Verificando QIIME 2...")

result = subprocess.run(
    ['conda', 'run', '-n', 'qiime2-amplicon-2025.10', 'qiime', '--version'],
    capture_output=True,
    text=True
)

if result.returncode == 0:
    version = result.stdout.strip()
    print(f"✅ {version}")
    print("   QIIME 2 está listo para análisis\n")
else:
    print("❌ QIIME 2 no está disponible")
    print("   Verifica que la instalación anterior haya completado correctamente\n")
```
